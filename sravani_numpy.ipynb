A1. Why are NumPy arrays generally preferred over plain Python lists for numerical computation in AI/ML workloads?

Answer:

NumPy arrays are better for numerical calculations because they are faster and use memory efficiently. They also support vectorized operations, so we can perform calculations on many values at once without writing loops. This makes NumPy useful for AI/ML tasks.

A2. Explain vectorization. Why does replacing an explicit Python loop with a vectorized NumPy operation matter with large datasets?

Answer:

Vectorization means performing an operation on an entire NumPy array at once instead of using a loop. It makes the code shorter and usually faster, especially when working with large amounts of data.

A3. For compatible 2D arrays A and B, explain the difference between A * B and A @ B. Give one neural-network situation where each is useful.

Answer:

A * B performs element-wise multiplication, so corresponding values are multiplied. A @ B performs matrix multiplication. In neural networks, * can be used for element-wise scaling or masking, while @ is commonly used to multiply inputs with weights.

A4. What is broadcasting in NumPy? State the general compatibility rule for broadcasting.

Answer:

Broadcasting allows NumPy to perform operations on arrays with different shapes. NumPy compares the dimensions from the last one, and the dimensions must either be equal or one of them must be 1.

A5. What is the difference between flatten() and ravel()? Which would you prefer for a large image when memory is constrained, and why?

Answer:

Both flatten() and ravel() convert an array into a 1D array. flatten() creates a copy, while ravel() may return a view and therefore use less memory. For a large image with limited memory, I would prefer ravel() when possible.

A6. Explain the difference between axis=0 and axis=1 in aggregation functions. Give a small example.

Answer:

In a 2D array, axis=0 performs the calculation down the rows, giving a result for each column. Axis=1 performs the calculation across the columns, giving a result for each row.

```python
#EXAMPLE:
import numpy as np

a = np.array([[1, 2, 3],
              [4, 5, 6]])

print(np.sum(a, axis=0))
print(np.sum(a, axis=1))
```

C1. Given a 1D NumPy array of exam scores, use np.where() to produce labels 'P' (score >= 40) or 'F' (score < 40). 

```python
import numpy as np

scores = np.array([35, 40, 55, 28, 75])

labels = np.where(scores >= 40, "P", "F")

print(labels)
```

Explanation: np.where() applies the pass/fail condition to all scores without using a loop

C2. Write a function standardize(x) that takes a 1D NumPy array and returns its z-score standardized version using only NumPy.

```python
def standardize(x):
    return (x - np.mean(x)) / np.std(x)

x = np.array([1, 2, 3, 4, 5])

print(standardize(x))
```

Explanation: The function subtracts the mean and divides by the standard deviation to standardize the values.

C3. Combine two 2D arrays (batches) with the same number of features using the appropriate stacking method. Print the shape of the combined array.


```python
a = np.array([[1, 2, 3],
              [4, 5, 6]])

b = np.array([[7, 8, 9],
              [10, 11, 12]])

combined = np.vstack((a, b))

print(combined)
print(combined.shape)
```

Explanation: np.vstack() combines the two batches row-wise because both arrays have the same number of features.

C4. Given X with shape (100, 5), compute the column-wise mean and standard deviation, then standardize X using a single vectorized expression.


```python
X = np.random.default_rng(42).normal(size=(100, 5))

mean = X.mean(axis=0)
std = X.std(axis=0)

X_standardized = (X - mean) / std

print("Mean:", mean)
print("Standard deviation:", std)
print("Standardized shape:", X_standardized.shape)
```

Explanation: axis=0 calculates the mean and standard deviation for each feature column, and broadcasting applies them to all rows.


C5. Given X, weights w, bias b, and true values y_true, compute the predictions using y_pred = X @ w + b and then calculate the Mean Squared Error (MSE).



```python
X = np.array([[1, 2],
              [2, 3],
              [3, 4]])

w = np.array([0.5, 1.0])
b = 2

y_true = np.array([4, 6, 8])

y_pred = X @ w + b
mse = np.mean((y_true - y_pred) ** 2)

print("Predictions:", y_pred)
print("MSE:", mse)
```

Explanation: Matrix multiplication calculates the predictions, and MSE measures the average squared difference between predicted and true values.

D1. Given a raw 1D or 2D NumPy array containing NaN values, write a step-by-step process to:
1. Detect missing values.
2. Handle them appropriately and justify your choice.
3. Normalize or standardize the resulting data.
4. Implement the full pipeline on a sample array containing NaN.


```python
import numpy as np

data = np.array([[10, 20, np.nan],
                 [30, np.nan, 50],
                 [40, 60, 70]])

# 1. Detect missing values
print("Missing values:")
print(np.isnan(data))

# 2. Replace NaN values with column means
column_means = np.nanmean(data, axis=0)
rows, cols = np.where(np.isnan(data))
data[rows, cols] = column_means[cols]

# 3. Standardize the data
mean = data.mean(axis=0)
std = data.std(axis=0)
standardized = (data - mean) / std

# 4. Display the result
print("Data after handling NaN:")
print(data)

print("Standardized data:")
print(standardized)
```

Explanation:

First, np.isnan() detects the missing values. The NaN values are replaced with their respective column means so that the available data is retained. Finally, the data is standardized using the column-wise mean and standard deviation.

X = np.array([[1, 2],
              [2, 3],

w = np.array([0.5, 1.0, 0.2])

Explanation:

There are two main bugs. First, X has shape (3, 2), so W must have 2 values, but the given W has 3 values. Second, the MSE formula needs the difference to be squared before taking the mean. The corrected expression is np.mean((y_true - y_pred) ** 2).

E1. Perform per-column min-max normalization on a 2D NumPy array. The solution must be fully vectorized and use no loops.



```python
import numpy as np

X = np.array([[10, 20, 30],
              [20, 40, 50],
              [30, 60, 70]], dtype=float)

X_min = X.min(axis=0)
X_max = X.max(axis=0)

X_normalized = (X - X_min) / (X_max - X_min)

print(X_normalized)
```

Explanation: axis=0 finds the minimum and maximum of each column, and broadcasting applies min-max normalization without using loops.
